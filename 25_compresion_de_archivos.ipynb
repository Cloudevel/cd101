# Paquetes de software.

Al existir tantas distribuciones basadas en la plataforma *GNU/Linux*, el tema de la distribución e instalación de paquetes de software se volvió un tema muy complejo.

La problemática aumenta en vista de que en la gran mayoría de los casos, un paquete de software requiere de una biblioteca de paquetes adicionales. El cálculo de las dependencias de un paquete fue durante mucho tiempo un gran obstáculo para la adopción de la plataforma.

## Compilación del código fuente.

La gran ventaja del software publicado bajo los términos de una licencia de software libre, es que se tiene acceso al código fuente de dicho software y dependiendo del lenguaje de programación en el que se encuentren codificados es posible crear archivos binarios que se ajusten a una plataforma específica.

Aún cuando esta forma de distribuir el software es muy poderosa, resulta muy complicada incluso para programadores y administradores de sistemas experimentados.

### El toolchain de GNU.

https://es.wikipedia.org/wiki/GNU_toolchain

### Pasos generales.

Por lo general, la forma en la que se compila el software es la siguiente:

* Configuración.
* Compilación.
* Instalación.

### Scripts de instalación.

Muchos desarrolladores han optado por incluir junto con el código fuente, guiones (scripts) escritos en lenguajes interpretados como Bash, Python, Perl, etc. los cuales automatizan la configuración y compilación del software que publica.


### Los archivos *README* e *INSTALL*.

En la mayoría de los casos, los desarrolladores incluyen archivos llamados *README* e *INSTALL* con diversas extensiones, los cual contienen los detalles e instrucciones de configuración, compilación e instalación del software que han publicado.

## Empaquetado y compresión de estructuras de archivos.

Por lo general, el software es publicado en *paquetes* que contienen estructuras de archivos y directorios, los cuales se utilizan para facilitar la descarga e instalación de dicho software.

### El comando ```tar```.

El comando ```tar``` es la herramienta más utilizada para empacar, comprimir y extraer archivos de las plataformas UNIX y GNU/Linux. Es un acrónimo de "Tape ARchiver".

A los archivos creados con ```tar``` se les conoce como *tarballs*, cuya traducción es *bolas de alquitrán*, en vista de quedichos archivos aglutinan a otros archivos. 

La sintaxis básica es la siguiente, pero cambia dependiendo de las opciones que se le añaden como argumentos.

```
tar <argumentos>
```

Las opciones básicas de ```tar``` son:
* ```-f``` o ```--file=<archivo>``` 
* ```-c``` o ```--create```
* ```-x``` o ```--extract``` o ```--get```
* ```-t``` o ```--list```
* ```-v``` o ```--verbose```
* ```-p``` o ```--preserve-permissions```
* ```-u``` o ```--update```
* ```-A``` o ```--catenate```
* ```-r``` o ```--append```
* ```-d``` o ```--diff```
* ```-C``` o ```--directory```

La siguiente celda prepara el directorio de práctica `data`, con tres archivos de texto pequeños, para que los ejemplos de `tar`, `gzip`, `bzip2` y `zip` tengan qué empaquetar. El directorio y los archivos que se generan en este cuaderno están en `.gitignore`, así que no se suben al repositorio.

````bash
mkdir -p data
for n in uno dos tres; do
    echo "Archivo de ejemplo $n" > "data/$n.txt"
done
ls -l data
````

La siguiente celda crea (`-c`) con `tar` el archivo (`-f`) `datos.tar`, que empaqueta el directorio `data` completo. `tar` solo agrupa: el archivo resultante no está comprimido.

````bash
tar -cf datos.tar data
````

La siguiente celda lista los elementos que coinciden con `*.tar`.


````bash
ls *.tar
````

La siguiente celda ejecuta `cp datos.tar ~/`.


````bash
cp datos.tar ~/
````

La siguiente celda lista los elementos que coinciden con `~/*.tar`.


````bash
ls ~/*.tar
````

La siguiente celda hará un listado de los contenidos del archivo.


````bash
tar -tvf ~/datos.tar
````

La siguiente celda extraerá un listado de los contenidos del archivo en el directorio actual.


````bash
mkdir datos_alternos
````

La siguiente celda ejecuta `tar -xvf ~/datos.tar -C datos_alternos`.


````bash
tar -xvf ~/datos.tar -C datos_alternos
````

La siguiente celda ejecuta `tree datos_alternos`.


````bash
tree datos_alternos
````

La siguiente celda ejecuta `touch data/nuevo`.


````bash
touch data/nuevo
````

La siguiente celda lista los elementos que coinciden con `data`.


````bash
ls data
````

La siguiente celda actualizará el archivo con respecto al directorio.


````bash
tar -uvf ~/datos.tar data
````

La siguiente celda ejecuta `tar -tvf ~/datos.tar`.


````bash
tar -tvf ~/datos.tar
````

### La compresión con ```gzip``` y ```gunzip```.

https://en.wikipedia.org/wiki/Gzip

 Opción ```-z```

La siguiente celda crea `datos.tar.gz`: `tar` empaqueta el directorio `data` y lo comprime con *gzip* (`-z`) en un solo paso; `-v` lista cada archivo agregado.

````bash
tar -cvzf datos.tar.gz data
````

### El comando ```gzip```.

Este comando permite realizar la compresión de cualquier archivo con el algoritmo *DEFLATE*.

La siguiente celda comprime con `gzip` el archivo `datos.tar` creado al inicio. `-k` conserva el original y `-f` sobrescribe el `datos.tar.gz` de la celda anterior; el resultado es equivalente. `gzip` comprime archivos individuales, no directorios: por eso se combina con `tar`.

````bash
gzip -kf datos.tar
````

#### La *man page* de ```gzip```.

La siguiente celda muestra la *man page* de `gzip`.


````bash
man gzip
````

### El comando ```gunzip```.

Este comando permite realizar la descompresión de un archivo con el algoritmo *DEFLATE*, el mismo que usa `gzip`.

#### La *man page* de ```gunzip```.

La siguiente celda muestra la *man page* de `gunzip`.


````bash
man gunzip
````

### La compresión ```bzip2``` y ```bunzip2```.

La compresión [*bzip2*](https://en.wikipedia.org/wiki/Bzip2) utiliza un algoritmo de compresión mejorado que permite generalmente crear archivos comprimidos más pequeños que ```zip```.

Opción ```-j```.

La siguiente celda crea `datos.tar.bz2` con `tar` y compresión *bzip2* (`-j`). Al compararlo con `ls -l`, suele ser más pequeño que el `.tar.gz`, a cambio de tardar más.

````bash
tar -cvjf datos.tar.bz2 data
````

La siguiente celda lista los elementos que coinciden con `-l *.tar.*`.


````bash
ls -l *.tar.*
````


### Los formatos *zip* y *rar*.

Estos formatos son sistemas basados en Windows. No son nativos de UNIX ni GNU/Linux, pero hay implementaciones.

Estos comandos no funcionan con tar.

#### ```zip``` y ```unzip```.

La siguiente celda empaqueta y comprime el directorio `data` en `datos.zip`. La opción `-r` recorre el directorio; sin ella, `zip` solo guardaría la entrada del directorio, vacía.

````bash
zip -r datos.zip data
````

La siguiente celda extrae `datos.zip` en el directorio `datos_zip` (`-d`). La opción `-o` sobrescribe sin preguntar, para que la celda no se quede esperando una respuesta.

````bash
unzip -o datos.zip -d datos_zip
````

El formato *rar* es otro formato que en la mayoría de los casos no viene preinstalado.

```rar``` y ``unrar``

La siguiente celda invoca `rar` para comprobar si está instalado. En el entorno del curso normalmente no lo está y la respuesta es `command not found`: *rar* es un formato propietario y su compresor no forma parte de las distribuciones por omisión.

````bash
rar
````

## Verificación de autenticidad.

Existen herramientas que permiten crear un *hash* -una secuencia de números única- que identifica al contenido de un archivo.

### ```sha*sum```.

La familia SHA (*Secure Hash Algorithm*) son una familia de algoritmos que permiten generar hash y son utilizadas para cifrado y autenticación.

https://es.wikipedia.org/wiki/Secure_Hash_Algorithm

La siguiente celda calcula con `sha1sum` el *hash* SHA-1 de cada archivo `.tar*`. Si un solo bit de un archivo cambia, su *hash* cambia por completo; así se comprueba que una descarga llegó íntegra.

````bash
sha1sum *.tar*
````

### ```md5sum```.

Este algoritmo fue muy utilizado en el pasado, sin embargo ha quedado en desuso debido a que se han logrado crear colisiones en el cálculo del hash.

La siguiente celda calcula con `md5sum` el *hash* MD5 de los mismos archivos. *MD5* y *SHA-1* ya no se consideran seguros frente a manipulaciones deliberadas; para verificar descargas se prefiere `sha256sum`.

````bash
md5sum *.tar*
````
