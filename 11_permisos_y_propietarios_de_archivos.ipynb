# Permisos y propietarios de archivos.

Tanto los directorios, archivos y procesos del sistema cuentan con dueños y conp permisos de acceso.

En este capítulo se explorará la forma de cambiar propietarios y permisos.

## Cambio de propietarios de recursos del sistema.

###  El comando ```chown```.

Este comando permite cambiar al propietario y grupo de un archivo o directorio.

```
chown <opciones> <usuario>:<grupo> <ruta>
```

#### Opciones de ```chown```.

* ```-R``` o ```--recursive```

La siguiente celda ejecuta `echo "<h1>Hola, mundo.</h1>" > ~/index.html`.


````bash
echo "<h1>Hola, mundo.</h1>" > ~/index.html
````

La siguiente celda lista los elementos que coinciden con `-l ~/index.html`.


````bash
ls -l ~/index.html
````

La siguiente celda cambia con `chown` el propietario de `~/index.html` al usuario `www-data` (el que usan los servidores web en *Ubuntu*). Solo `root` puede regalar archivos a otro usuario; por eso hace falta `sudo`.

````bash
sudo chown www-data ~/index.html
````

La siguiente celda lista los elementos que coinciden con `-l ~/index.html`.


````bash
ls -l ~/index.html
````

La siguiente celda ejecuta `sudo chown www-data:root ~/index.html`.


````bash
sudo chown www-data:root ~/index.html
````

La siguiente celda lista los elementos que coinciden con `-l ~/index.html`.


````bash
ls -l ~/index.html
````

#### La *man page* de ```chown```.

La siguiente celda muestra la *man page* de `chown`.


````bash
man chown
````

### El comando ```chgrp```.

Este comando permite cambiar al grupo al que pertenece un directorio o archivo.

```
chgrp <opciones> <grupo> <ruta>
```

La siguiente celda cambia con `chgrp` el grupo propietario de `~/index.html` a `www-data`. El resultado se comprueba con `ls -l`, en la cuarta columna.

````bash
sudo chgrp www-data ~/index.html
````

La siguiente celda lista los elementos que coinciden con `-l ~/index.html`.


````bash
ls -l ~/index.html
````

La siguiente celda elimina `~/index.html` con `rm -f`, después de usarlo en las celdas anteriores. La opción `-f` (*force*, forzar) borra el archivo sin pedir confirmación, incluso si está protegido contra escritura porque su propietario ya no es el usuario actual. La opción `-i` (*interactive*, interactivo) pediría confirmación antes de borrar.


````bash
rm -f ~/index.html
````

#### La *man page* de ```chgrp```.

La siguiente celda muestra la *man page* de `chgrp`.


````bash
man chgrp
````

## El comando ```chmod```.

Este comando permite modificar los permisos de acceso de un directorio o archivo.

```
chmod <opciones> <permisos> <ruta>
```

La siguiente celda lista los elementos que coinciden con `-l ~`.


````bash
ls -l ~
````

La siguiente celda ejecuta `touch archivo`.


````bash
touch archivo
````

La siguiente celda lista los elementos que coinciden con `-l archivo`.


````bash
ls -l archivo
````

La siguiente celda asigna con `chmod` los permisos `764` en notación octal: `7` (lectura, escritura y ejecución) para el propietario, `6` (lectura y escritura) para el grupo y `4` (solo lectura) para los demás.

````bash
chmod 764 archivo
````

La siguiente celda lista los elementos que coinciden con `-l archivo`.


````bash
ls -l archivo
````

La siguiente celda ejecuta `chmod 1762 archivo`.


````bash
chmod 1762 archivo
````

La siguiente celda lista los elementos que coinciden con `-l archivo`.


````bash
ls -l archivo
````

La siguiente celda ejecuta `chmod +x archivo`.


````bash
chmod +x archivo
````

La siguiente celda lista los elementos que coinciden con `-l archivo`.


````bash
ls -l archivo
````

La siguiente celda ejecuta `chmod -x archivo`.


````bash
chmod -x archivo
````

La siguiente celda lista los elementos que coinciden con `-l archivo`.


````bash
ls -l archivo
````

### La *man page* de ```chmod```.

La siguiente celda muestra la *man page* de `chmod`.


````bash
man chmod
````
