# Gestión de servicios con ```systemd```.

[*Systemd*](https://es.wikipedia.org/wiki/Systemd) es el sistema de inicio del sistema más común en las distribuciones de *GNU/Linux*. Es el primer proceso que se ejecuta una vez que el *kernel* termina de cargarse correctamente.

La labor de los sistemas de inicio como *init* de los sistemas *UNIX* basados en *System V* y *Systemd* es la de iniciar, configurar y gestionar los diversos daemons y procesos que ejecuta in sistema.

## El directorio ```/lib/systemd/system```.

En este directorio se encuentran todos los servicios que gestiona *Systemd*.

La siguiente celda enlistará el contenido del directorio ```/lib/systemd/system```.


````bash
sudo ls /lib/systemd/system
````

## Unidades de *Systemd*.

Las unidades son documentos de texto que describen el funcionamiento de diversas tareas gestionadas por *Systemd*. 

### Los ```targets``` de *Systemd*.

Los ```targets``` de *Systemd* son unidades que agrupan una serie de operaciones para alcanzar un estado específico del sistema. Se identifican debido a que tiene la extensión ```.target```.

La siguiente celda ejecuta `sudo ls /lib/systemd/system/*.target`.


````bash
sudo ls /lib/systemd/system/*.target
````

### Runlevels.

Un *runlevel* define la forma en la que el entorno de un sistema *UNIX* o *GNu/Linux* se presenta a los usuarios.

* ```0``` Apagado.
* ```1``` Modo de rescate. 
* ```2``` Definido por el usuario.
* ```3``` Multi-usuario en modo de texto.
* ```4``` Definido por el usuario.
* ```5``` Multi-usuario con entorno gráfico.
* ```6``` Reinicio del sistema.

### El comando ```runlevel```.

Este comando indica el número del runlevel en el que se encuentra el sistema y si ha sido modificado desde el arranque.

La siguiente celda ejecuta `runlevel`.


````bash
runlevel
````

La siguiente celda lista los elementos que coinciden con `-al /lib/systemd/system/runlevel*.target`.


````bash
ls -al /lib/systemd/system/runlevel*.target
````

La siguiente celda muestra el contenido de `/lib/systemd/system/runlevel5.target`.


````bash
cat /lib/systemd/system/runlevel5.target
````

La siguiente celda muestra el contenido de `/lib/systemd/system/multi-user.target`.


````bash
cat /lib/systemd/system/multi-user.target
````

### Servicios.

Los servicios son unidades que permiten habilitar un servicio del sistema. Se identifican por la extensión ```.service```.

La siguiente celda lista los elementos que coinciden con `/lib/systemd/system/*.service`.


````bash
ls /lib/systemd/system/*.service
````

Los archivos de unidad de *systemd* suelen estar en `/lib/systemd/system/` o `/usr/lib/systemd/system/`. La celda previa lista los que hay en este entorno, si el directorio existe. En *GitHub Codespaces* la lista puede ser limitada.


La siguiente celda comprueba si existe una unidad `cron.service` en esas rutas (solo consulta archivos; no usa `systemctl`).


````bash
ls /lib/systemd/system/cron.service /usr/lib/systemd/system/cron.service 2>/dev/null || echo "No hay una unidad cron.service en este entorno."
````

## Comandos de gestión de servicios.

### ```systemctl```.
Es el comando de gestión de los servicios de systemd


``` 
systemctl <acción> <servicio>

```

#### Acciones.

* ```enable```
* ```disable```
* ```start```
* ```status```
* ```stop```

**Nota sobre el entorno:** las celdas siguientes usan ```systemctl``` para consultar, detener y arrancar servicios, con el servidor web *Apache* (```apache2```) como ejemplo. Requieren una máquina *Ubuntu* 24.04 con *systemd* como *PID 1* y *Apache* instalado (```sudo apt install apache2```). En *GitHub Codespaces* *systemd* no es el *PID 1* y *Apache* no está instalado, así que ```systemctl``` responde ```System has not been booted with systemd as init system (PID 1). Can't operate.``` Por la misma razón, ```runlevel``` no informa un nivel válido en *Codespaces*.

La siguiente celda consulta el estado de un servicio con `systemctl`.


````bash
sudo systemctl status --no-pager
````

La siguiente celda consulta el estado de un servicio con `systemctl`.


````bash
systemctl status apache2.service --no-pager
````

La siguiente celda ejecuta `sudo systemctl stop apache2`.


````bash
sudo systemctl stop apache2
````

La siguiente celda consulta el estado de un servicio con `systemctl`.


````bash
sudo systemctl status apache2 --no-pager
````

La siguiente celda ejecuta `sudo systemctl start apache2`.


````bash
sudo systemctl start apache2
````

La siguiente celda consulta el estado de un servicio con `systemctl`.


````bash
sudo systemctl status apache2 --no-pager
````

###  El comando ```service```.

El comando *service* es el comando usado por *System V* para la gestión de servicios.

Algunas distribuciones de GNU/Linux enmascaran *systemd* y traducen sus sinstrucciones mediante *service*.

```
service <servicio> <acción>
```

* La siguiente instrucción deberá ser ejecutada desde una terminal.

```
sudo service apache2 status
```
