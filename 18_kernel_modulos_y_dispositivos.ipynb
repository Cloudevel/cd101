# El kernel de *Linux*, módulos y dispositivos.

## Archivos del kernel de Linux.

La arquitectura del kernel de Linux permite tener un componente "monolítico" al cual se le pueden añadir módulos de forma dinámica.

### El componente monolítico de kernel.

El componente monolítico, del kernel contiene los elementos críticos de un sistema y  por lo general es un archivo ejecutable cuyo nombre inicia con [*vmlinuz*](http://www.linfo.org/vmlinuz.html) y se encuentra en el directorio */boot/* del sistema.

La siguiente celda lista los elementos que coinciden con `/boot/vmlinuz*`.


````bash
ls /boot/vmlinuz*
````

## El comando ```uname```.

Ese comando regresa la información básica del sistema, incluyendo la versión actual del kernel.

```
uname <opciones>
```
### Algunas opciones de ```uname```.

* ```-a```o ```--all``` regresa todos los datos del sistema.
* ```-s```o ```--kernel-name``` regresa el nombre del kernel.
* ```-r```o ```--kernel-release``` regresa la versión del kernel de Linux.
* ```-v```o ```--kernel-version``` regresa la versión de la compilación del kernel.
* ```-m```o ```--machine``` regresa los datos del procesador para el que fue compilado.
* ```-o```o ```--operating-system``` regresa todos los datos del sistema operativo para el que fue compilado.

La siguiente celda ejecuta `uname`, que muestra el nombre del *kernel* (`Linux`). Con `-r` muestra su versión y con `-a` toda la información disponible.

````bash
uname
````

La siguiente celda ejecuta `uname -a`.


````bash
uname -a
````

La siguiente celda ejecuta `uname -s`.


````bash
uname -s
````

La siguiente celda ejecuta `uname -r`.


````bash
uname -r
````

La siguiente celda ejecuta `uname -v`.


````bash
uname -v
````

La siguiente celda ejecuta `uname -m`.


````bash
uname -m
````

La siguiente celda ejecuta `uname -o`.


````bash
uname -o
````

### La man page de ```uname```.

La siguiente celda muestra la *man page* de `uname`.


````bash
man uname
````

### Los módulos del kernel.

Además del componente monolítico, el sistema puede añadir "módulos" al kernel, los cuales están en el directorio ```/lib/modules```. Por lo general, el directorio en el que se encuentra corresponde a ```uname -r```.

La siguiente celda lista los elementos que coinciden con `/lib/modules/$(uname -r)`.


````bash
ls /lib/modules/$(uname -r)
````

## Listado de módulos con ```lsmod```. 

La siguiente celda lista con `lsmod` los módulos del *kernel* cargados, su tamaño y cuántos módulos los usan. La información proviene de `/proc/modules`.

````bash
lsmod
````

## Desinstalación de módulos ```rmmod``` .

**Advertencia:** desinstalar un módulo del *kernel* (`rmmod`) puede dejar inservibles dispositivos en uso. En *GitHub Codespaces* el *kernel* es el del anfitrión. No ejecute `rmmod` en un sistema de producción ni sobre módulos que no controla.


La siguiente celda ejecuta `echo "Comando omitido: véase la advertencia sobre rmmod."`.


````bash
# sudo rmmod snd
echo "Comando omitido: véase la advertencia sobre rmmod."
````

La siguiente celda muestra la *man page* de `rmmod`.


````bash
man rmmod
````

## Gestión de módulos con ```modprobe ```.

La siguiente celda ejecuta `modprobe snd`.


````bash
modprobe snd
````

#### El directorio */etc/modprobe.d/*

Este directorio contiene archivos de configuración para la gestión de módulos.

La siguiente celda lista los elementos que coinciden con `/etc/modprobe.d`.


````bash
ls /etc/modprobe.d
````

La siguiente celda muestra el contenido de `/etc/modprobe.d/blacklist-ath_pci.conf`.


````bash
cat /etc/modprobe.d/blacklist-ath_pci.conf
````

La siguiente celda muestra el contenido de `/etc/modprobe.d/iwlwifi.conf`.


````bash
cat /etc/modprobe.d/iwlwifi.conf
````

La siguiente celda muestra la *man page* de `modprobe`.


````bash
man modprobe
````

## Cálculo de dependencias de módulos con ```depmod```.




## Gestión de módulos.
### Detección con ```udev```.

El daemon *udev* es el sistema encargado de la detección de dispositivos, controladore y módulos.

## El directorio ```/dev/```.

El directorio ```/dev/``` contiene referencia a todos los dispositivos y módulos del sistema.

La siguiente celda lista los elementos que coinciden con `/dev/`.


````bash
ls /dev/
````

## Manipulación de parámetros del kernel con _sysctl_.

El comando sysctl es un un componente de manipulación avanzada del kernel a partir de ```systemd```.

## Información sobre dispositivos.

###  ```lshw```
### ```lsusb```.
### ```lspci```.
### ```lscpu```.

La siguiente celda describe con `lshw` el *hardware* del equipo en forma de árbol. Necesita `sudo` para obtener la información completa; en un contenedor solo ve parte del anfitrión.

````bash
sudo lshw
````

````bash
sudo lsusb
````

La siguiente celda lista con `lspci` los dispositivos conectados al bus *PCI*: controladores de red, de video, de almacenamiento, etc.

````bash
lspci
````

La siguiente celda resume con `lscpu` la arquitectura del procesador: modelo, número de núcleos e hilos, cachés y banderas.

````bash
lscpu
````
