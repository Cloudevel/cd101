# La línea de comandos.

## Terminales.

Tanto *GNU/Linux* como *UNIX* se basan en la arquitectura *cliente-servidor*, la cual permite a múltiples usuarios conectarse y acceder de forma simultánea a un servidor mediante una o más *sesiones*. 

A la interfaz de conexión de un cliente a un servidor *GNU/Linux*, se le conoce como *terminal*.


### Terminales de texto.

La terminal de texto, también conocida como consola,  es la interfaz básica de un sistema *UNIX* y *GNU/Linux*. 

Las terminales de texto emulan a una pantalla que despliega caracteres de texto y por lo general requieren de un usuario y contraseña válida para iniciar una sesión en el servidor al que se conectan. 

La interfaz de entrada es por medio del teclado y la de salida se realiza mediante el despliegue de caracteres en la pantalla.

### Terminales gráficas.

Tanto los sistemas *UNIX* modernos como los sistemas *GNU/Linux* permiten gestionar terminales gráficas por medio de servidores especializados en el despliegue de dichas interfaces. 

Un sistema de este tipo cuenta con un servidor de entorno gráfico al cual pueden conectarse diversos clientes. 

El [*Sistema de ventanas X*](https://es.wikipedia.org/wiki/Sistema_de_ventanas_X) o simplemente *las X* es el sistema más popular en *UNIX* y *GNU/Linux*, mientras que *MacOS X* utiliza [*Quartz*](https://en.wikipedia.org/wiki/Quartz_(graphics_layer)).

### Entornos de escritorio de *GNU/Linux*.

Las *X* sólo proporcionan un servicio básico de despliegue de elementos gráficos y de ventanas simples, por lo que junto con estas, se utilizan gestores de entornos de escritorio, los cuales permiten correr diversas aplicaciones.

Los entornos gráficos más populares son:
* [*KDE* (*K Desktop Environment*)](https://kde.org/)
* [*GNOME* (*GNU Network Object Model Environment*)](https://gnome.org).
* [*Mate*](https://mate-desktop.org/).
* [*Xfce*](https://xfce.org/).
* [*AfterStep*](http://afterstep.org/).

**Nota:** Este curso tiene por objetivo aprovechar las capacidades de las terminales de texto.

## El *shell*.

La terminal o consola es  únicamente la interfaz que gestiona la entrada y salida de una conexión. Cuando un usuario inicia una sesión en una terminal de texto, esta por lo general le asigna un entorno interactivo que se conoce como *shell*.

Un *shell* permite interactuar con el servidor mediante comandos predefinidos y también ejecutando aplicaciones del sistema. 

Existen varios shell para *UNIX* y *GNU/Linux*:

* [*C shell*](https://en.wikipedia.org/wiki/C_shell) (csh y tcsh).
* [*Zshell*](https://en.wikipedia.org/wiki/Z_shell) (zsh).
* [*Korn shell*](https://en.wikipedia.org/wiki/KornShell)(ksh).
* [*Bourne shell*](https://en.wikipedia.org/wiki/Bourne_shell) (sh).
* [*Bourne-again shell*](https://en.wikipedia.org/wiki/Bash_(Unix_shell)) (bash).

En este curso se utilizará *Bash*, el cual es el *shell* más común en distribuciones de *GNU/Linux*.

## El *prompt*.

Una vez que se ha iniciado una sesión ingresando los datos de un usuario y una contraseña válidos, la terminal desplegará una serie de caracteres a la izquierda de una línea de inicio y a su derecha un cursor parpadeando. A esto se le conoce como el *prompt*.

## Ejecución de comandos.

Los comandos de una terminal tiene la siguiente estructura:

```
<comando> <subcomando> <opciones> <argumentos>
```

### Ejecución de comandos en las notebooks de *Jupyter*.

Las notebooks de *Jupyter* utilizadas en este curso no son terminales, sino interfaces que se conectan a un *shell de bash*.

Debido a lo anterior, las celdas de código de las notebooks pueden ejecutar comandos de *bash* de forma directa.

**ADVERTENCIA:** Las celdas sólo pueden enviar comandos y desplegar el resultado de estos comandos. Si existe algún tipo de interacción adicional por parte del shell, la celda se quedará bloqueada y será necesario interrumpir el kernel para salir de ese estado.

**Ejemplos:**

* El siguiente comando indicará el directorio en el que se encuentra esta notebook.

La siguiente celda ejecuta `pwd`.


````bash
pwd
````

La siguiente celda traerá un listado completo del directorio *home* del usuario corriendo *Jupyter*, incluyendo los archivos escondidos.


````bash
ls ~ -al
````

## Cómo obtener ayuda.

Existen ciertos comandos que nos permiten buscar ayuda sobre otros comandos del sistema.

### Las *man pages*.

Una de las mejores prácticas al desarrollar aplicaciones en *GNU/Linux* y en *UNIX* es que por lo general, cada comando contiene documentación detallada en forma de manual a la que se conoce como *man page*.

Las *man page* describen el funcionamiento de un comando y todas las opciones y parámetros que puede aceptar, incluyendo una breve descripción.

### El comando ```man```.

Para acceder a la *man page* de un comando se utiliza el comando ```man```.

``` 
man <comando>
```

Este comando desplegará la *man page* en una ventana que le permite al usuario desplazarse por el texto utilizando las teclas:
* <kbd>Arriba</kbd>
* <kbd>Abajo</kbd>
* <kbd>Página Arriba</kbd>
* <kbd>Página Abajo</kbd>
* <kbd>Inicio</kbd>
* <kbd>Fin</kbd>

Para obtener ayuda se presiona la tecla <kbd>h</kbd>.

Para salir de la ventana de ayuda se presiona la tecla <kbd>q</kbd>.

**Ejemplos:**

* En el caso de la siguiente celda, se abrirá una la ventana con la *man page* del comando ```ls``` que puede ser vista usando el desplazamiento del mouse o la barra de desplazamiento lateral.

La siguiente celda muestra la *man page* de `ls`.


````bash
man ls
````

### El comando ```whatis```.

Este comando ```whatis``` trae sólo la descripción del comando buscado.

```whatis <comando>```

## El comando ```apropos```.

El comando ```apropos``` regresa una lista de comandos que incluyan el término indicado en la descripción de su *man page*.

```
apropos <término>
```

**Ejemplo:**

*


La siguiente celda traerá un listado de comandos que contengan el término ```network``` en su *man page*.


````bash
apropos network
````

### El comando ```whereis```.

Este comando regresa la ruta en la que se encuentran los comandos, las bibliotecas y la configuración que correspondan a un comando en particular.

```
whereis <término>
```

**Ejemplo:**

*


La siguiente celda desplegará la localización del término ```python```.


````bash
whereis python
````

### El comando ```which```.

Este comando indica la ruta en la que se encuentra un comando específico.

```
which <comando>
```

**Ejemplo:**

*


La siguiente celda desplegará la ruta donde se encuentra el comando ```python3```.


````bash
which python3
````

## Cierre de la sesión.

Para cerrar la sesión en una terminal se utiliza el comando ```exit```.
