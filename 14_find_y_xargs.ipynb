# Los comandos *find* y *xargs*.

## El comando ```find```.

Este comando permite realizar diversos tipos de búsquedas a partir de un directorio por medio de una expresión.

```
find <directorio inicial> <expresión>
```

Las expresiones pueden estar conformadas por:

* Opciones.
* Pruebas.
* Acciones.

La siguiente celda busca con `find`, a partir del directorio actual (`.`), los archivos cuyo nombre (`-name`) termina en `.ipynb`. El patrón va entre comillas para que lo interprete `find` y no el *shell*; la salida es una ruta por línea, relativa al directorio de partida.

````bash
find . -name "*.ipynb"
````

### Algunas opciones.

* ```-L``` continúa la búsqueda en los directorios con ligas simbólicas.
* ```-P``` no realiza la búsqueda en los directorios con ligas simbólicas.
* ```-maxdepth``` indica el máximo número de niveles de subdirectorios en los que se realizará la búsqueda.
* ```-mindepth```  indica el número mínimo de niveles de subdirectorios en los que se realizará la búsqueda.

La siguiente celda ejecuta `find . *.ipynb -maxdepth 1`.


````bash
find . *.ipynb -maxdepth 1
````

La siguiente celda ejecuta `find . *.ipynb -maxdepth 2`.


````bash
find . *.ipynb -maxdepth 2
````

### Criterios numéricos.

* ```+n``` mayor que n.
* ```-n``` menor que n.
* ```n``` igual a n.

## El comando ```xargs```.

Este comando es un filtro que usa como argumento para un comando cada elemento de un flujo de datos.

```
 <pipe> xargs <comando> 
```

La siguiente celda lista el contenido del directorio de trabajo.


````bash
ls
````

La siguiente celda lista los elementos que coinciden con `| xargs wc`.


````bash
ls | xargs wc
````

La siguiente celda muestra la *man page* de `xargs`.


````bash
man xargs
````
