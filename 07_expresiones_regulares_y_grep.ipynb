# Expresiones regulares y *grep*.

## Expresiones regulares y comodines del *shell*.

Las *expresiones regulares* forman parte de los lenguajes formales y permiten identificar patrones de texto mediante una combinación de caracteres especiales. Las usan, entre otros, `grep`, `sed` y `gawk`.

Los *comodines* del *shell* (*globs* o *filename expansion*) son un mecanismo distinto: el *shell* los expande a nombres de archivos *antes* de ejecutar el comando. El comando `ls` no interpreta expresiones regulares; recibe ya la lista expandida.

No se deben mezclar las dos notaciones. En un *glob*, `?` equivale a un carácter y `*` a cualquier cadena (incluida la vacía). En una expresión regular, `.` equivale a un carácter, `*` repite el elemento anterior y `?` (en *ERE* (expresiones regulares extendidas, *Extended Regular Expressions*), con `grep -E`) lo hace opcional.


### Caracteres especiales de las expresiones regulares.

La tabla siguiente describe la notación habitual en `grep`. Los operadores `?`, `+`, `()` y `{}` requieren expresiones regulares extendidas (*ERE*), que se activan con `grep -E`.

| Carácter | Acción |
|:---:|:-------|
| `^` | Ancla el patrón al principio de una línea. |
| `$` | Ancla el patrón al final de una línea. |
| `.` | Coincide con un carácter cualquiera. |
| `*` | Repite el elemento anterior cero o más veces. |
| `?` | En *ERE*, el elemento anterior es opcional (cero o una vez). |
| `+` | En *ERE*, repite el elemento anterior una o más veces. |
| `\` | Carácter de escape. |
| `[` ... `]` | Coincide con un carácter del conjunto o intervalo indicado. |
| `[^` ... `]` | Coincide con un carácter que no está en el conjunto. |
| `(` *expr1* \| *expr2* `)` | En *ERE*, agrupa alternativas. |
| `{n}` | En *ERE*, el elemento anterior se repite exactamente *n* veces. |
| `{n,m}` | En *ERE*, el elemento anterior se repite entre *n* y *m* veces. |


## Comodines del *shell* con el comando `ls`.

El *shell* expande los comodines a nombres de archivos que coinciden con el patrón. La forma habitual es:

```
ls <ruta/><glob>
```

| Comodín | Acción |
|:---:|:-------|
| `*` | Cualquier cadena de caracteres, incluida la vacía. |
| `?` | Exactamente un carácter. |
| `[abc]` | Un carácter del conjunto indicado. |
| `[0-9]` | Un carácter del intervalo indicado. |
| `[!abc]` | Un carácter que no pertenece al conjunto (en *Bash*). |

Las celdas siguientes ilustran esta expansión. Por ejemplo, `ls *.ipynb` lista los archivos cuyo nombre termina en `.ipynb`, y `ls 1?_*` lista los que tienen un carácter entre `1` y `_`.


La siguiente celda lista los nombres que coinciden con el comodín `*`.


````bash
ls *
````

La siguiente celda lista los archivos con extensión `.ipynb`.


````bash
ls *.ipynb
````

La siguiente celda lista los elementos que coinciden con `1?_*`.


````bash
ls 1?_*
````

La siguiente celda lista los elementos que coinciden con `1[1-5]*`.


````bash
ls 1[1-5]*
````

La siguiente celda lista los elementos que coinciden con `?[1-5]*`.


````bash
ls ?[1-5]*
````

La siguiente celda lista los elementos que coinciden con `[02][^1-5]*`.


````bash
ls [02][^1-5]*
````

La siguiente celda lista los elementos que coinciden con `1[370]*`.


````bash
ls 1[370]*
````

La siguiente celda lista los elementos que coinciden con `*ft*`.


````bash
ls *ft*
````

La siguiente celda lista los elementos que coinciden con `*[xf][tp]*`.


````bash
ls *[xf][tp]*
````

## La familia de comandos ```grep```.

Para aprovechar el uso de expresiones regulares en flujos de texto existe una familia de comandos que buscan las líneas que cumplen con una expresión regular. El comando más utilizado de esta familia es ```grep```.

```
grep <opciones> <regexp> <flujo>
```

### Opciones.

* ```-i``` permite que las mayúsculas y minúsculas sean indistintas.
* ```-E``` interpreta el patrón como expresión regular extendida (*ERE*).


La siguiente celda muestra el contenido de `LICENSE`.


````bash
cat LICENSE
````

La siguiente celda muestra con `grep` las líneas de `LICENSE` que contienen la cadena `Attribution`. El archivo está bajo la licencia Creative Commons Atribución 4.0 Internacional (*CC BY 4.0*). La búsqueda distingue mayúsculas de minúsculas; `-i` desactiva esa distinción.

````bash
grep Attribution LICENSE
````

La siguiente celda busca con `grep` la cadena `CC BY` en `LICENSE`. El texto legal usa el nombre completo *Creative Commons Attribution 4.0 International*, no esa abreviatura, de modo que no hay coincidencias.


````bash
grep "CC BY" LICENSE
````

La siguiente celda busca con `grep -i` la cadena `attribution` en `LICENSE`. Con `-i`, las mayúsculas y las minúsculas son indistintas, así que también aparecen las líneas que contienen `Attribution`.


````bash
grep -i attribution LICENSE
````

La siguiente celda busca con `grep` las líneas de `LICENSE` que empiezan por `commons`, sin distinguir mayúsculas de minúsculas. El acento circunflejo `^` ancla el patrón al principio de la línea. La palabra aparece en el archivo, pero ninguna línea empieza por ella, así que no hay coincidencias.


````bash
grep -i ^commons LICENSE
````

La siguiente celda busca coincidencias con `grep`.


````bash
grep -i r$ LICENSE
````

La siguiente celda busca coincidencias con `grep`.


````bash
grep -i ^the LICENSE
````

La siguiente celda busca coincidencias con `grep`.


````bash
grep -Ei "y*\e{2}" LICENSE
````

La siguiente celda busca coincidencias con `grep`.


````bash
grep -Ei r\{2} LICENSE
````

La siguiente celda busca coincidencias con `grep`.


````bash
grep -Ei "(license|liability)" LICENSE
````

La siguiente celda busca coincidencias con `grep`.


````bash
grep -Ei "(license|merge)" LICENSE
````

La siguiente celda muestra con `dmesg` los mensajes del *kernel* (detección de *hardware*, controladores, errores).

La lectura de estos mensajes puede estar restringida. Ocurre en contenedores sin privilegios, porque el *kernel* pertenece al anfitrión, y en sistemas donde el parámetro `kernel.dmesg_restrict` vale `1`, que solo permite leerlos con privilegios de administrador. En ese caso `dmesg` responde `Operation not permitted`. Por eso la celda lo intenta primero sin privilegios, después con `sudo -n` (sin pedir contraseña) y, si tampoco es posible, muestra un aviso.

````bash
dmesg || sudo -n dmesg || echo "Este entorno no permite leer los mensajes del kernel."
````

La siguiente celda filtra con `grep -i "input"` los mensajes del *kernel* sobre dispositivos de entrada, con la misma alternativa de `sudo -n`. Si el entorno no permite leer los mensajes o ninguno coincide, `grep` termina con un código de error.

````bash
{ dmesg 2>/dev/null || sudo -n dmesg; } | grep -i "input"
````
